```python
import os
from dotenv import load_dotenv

load_dotenv()
api_key=os.environ.get("GROQ_API_KEY")
```

```python
#LLM Building
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0.7
)
```

```python
#Messages

from langchain.messages import HumanMessage , SystemMessage

response = llm.invoke([
    SystemMessage(content="You are a movie recommendation system."),
    HumanMessage(content="Suggest me best 5 Salman Khan movies. Give me just names not any explanation")
])

print(response.content)
```
