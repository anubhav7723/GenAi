## LLM USAGE

```python
#LLM Building
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0.7
)
```

```python
llm1 = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0
)
llm2 = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 1
)
```

```python
response1 = llm1.invoke([
    SystemMessage(content="You are a food recommendation system."),
    HumanMessage(content="Suggest me best 5 italian food")
])

response2 = llm2.invoke([
    SystemMessage(content="You are a food recommendation system."),
    HumanMessage(content="Suggest me best 5 italian food")
])



print(response1.content)

print(response2.content)
```

```python
#Messages

from langchain.messages import HumanMessage , SystemMessage

response = llm.invoke([
    SystemMessage(content="You are a movie recommendation system."),
    HumanMessage(content="Suggest me best 5 Akshay Kumar movies. Give me just names not any explanation")
])

print(response.content)
```

```python
# Streaming Responses
for chunk in llm.stream("Explain Generative AI"):
    print(chunk.content, end="")
```

```python
# Batch processing -> useful for large scale generation task

responses = llm.batch([
    "Explain Ai in just 5 lines.",
    "Explain ML in just 5 lines."
])
print([response.content for response in responses])
```

## Prompt Engineering

```python
# zero shot prompting -> ask que without giving examples

response = llm.invoke([
    HumanMessage(content="Classify this dish as veg or non-veg: Mutton Biryani")
])
print(response.content)
```

```python
# few shot prompting -> giving examples before asking question

response = llm.invoke([
    HumanMessage(content="""
Dish: Paneer Tikka
Category: Vegetarian

Dish: Chicken Biryani
Category: Non-Vegetarian

Dish: Sushi
Category:
""")
])

print(response.content)
```

```python
# Role prompting -> assinging a role to AI before asking anything

response = llm.invoke([
    SystemMessage(content="You are a HealthCare Assistant."),
    HumanMessage(content="Suggest name of medicines to recover from fever. Just give me name of medicine")
])

print(response.content)
```

```python
# Chain of Thought prompting -> Think step by step before answering.

response = llm.invoke([
    HumanMessage(content="I sold 50 pizzas today and each cost is 150 rupees. What is my today revenue? Think step by step")
])
print(response.content)
```

```python
# Prompt Templates (Langchain)
from langchain_core.prompts import ChatPromptTemplate

prompt =  ChatPromptTemplate.from_template(
    "Suggest 5 {cuisine} dishes. Just name."
)

#using chain

chain = prompt | llm

response = chain.invoke({
    'cuisine':'indian'
})

print(response.content)
```

```python
# Structured Prompthing -> want to get answer in specific format

response = llm.invoke([
    HumanMessage(content="suggest me 5 indian dishes and return ans in CSV format  with name and one main ingredient")
])
print(response.content)
```

```python
#OverALL

response = llm.invoke([
    SystemMessage(content="You are a restaurant recommendation assistant."),
    HumanMessage(content="suggest me 5 indian dishes and return ans in CSV format  with name and one main ingredient")
])
print(response.content)
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

documents = [
    "Pizza is an Italian dish",
    "Pasta is made from wheat",
    "Sushi is Japanese food"
]

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_texts(documents, embeddings)
```

```python
docs = vectorstore.similarity_search("What food is Italian?",k=2)

print(docs[0].page_content, " | " , docs[1].page_content)
```

```python
retriever = vectorstore.as_retriever()
docs = retriever.invoke("italian")
print(docs[0].page_content)
```

## Complete RAG pipeline

```python
documents = [
    "Pizza is a famous Italian dish made with cheese and tomato sauce.",
    "Pasta is a staple Italian food usually made from wheat.",
    "Sushi is a traditional Japanese dish made with rice and seafood.",
    "Anubhav is 3rd year Btech student",
    "He is pursuing his degree from UEM jaipur."
]
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.create_documents(documents)
print(chunks)
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
texts = [doc.page_content for doc in chunks]

vectorstore = FAISS.from_texts(texts, embeddings)

retriever = vectorstore.as_retriever()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

prompt = ChatPromptTemplate.from_template(
    '''
    You are a helpful assistant.

    Answer the question using the provided context.

    Context:
    {context}

    Question:
    {question}

    If this information is not sufficient then answer in general.
    '''
)

parser = StrOutputParser()

rag_chain = (
    {"context": retriever , "question":RunnablePassthrough()} | prompt | llm | parser
)
```

```python
response = rag_chain.invoke("tell me something about MicroSoft & Anubhav")
print(response)
```
